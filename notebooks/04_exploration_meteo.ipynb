# Exploration Open-Meteo — Historical Weather API

limites (dates inversées, dates futures, coordonnées invalides).

orchestration) et `src/cleaning/meteo.py` (extraction), avec tests
unitaires associés (`tests/test_ingestion_meteo.py`,
`tests/test_cleaning_meteo.py`).

Ce notebook sert maintenant d'exemple d'usage des modules.
