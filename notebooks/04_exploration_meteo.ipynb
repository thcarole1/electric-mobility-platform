# Exploration Open-Meteo — Historical Weather API

Exploration de l'API /v1/archive pour préparer l'ingestion météo,
destinée à être croisée avec les données Open Charge Map déjà extraites.

```python
import requests

url = "https://archive-api.open-meteo.com/v1/archive"
params = {
    "latitude": 48.856614,
    "longitude": 2.352222,
    "start_date": "2026-07-20",
    "end_date": "2026-07-24",
    "hourly": "temperature_2m",
}

response = requests.get(url, params=params)
print(response.status_code)
print(response.json())
```

```python

donnees = response.json()
temps = donnees["hourly"]["time"]
temperatures = donnees["hourly"]["temperature_2m"]
resultat = []

for t, temp in zip(temps, temperatures):
    resultat.append({"time": t, "temperature_2m" : temp})
    
resultat
```

```python
temps = ["2026-07-20T00:00", "2026-07-20T01:00"]
temperatures = [15.5, 14.9]

for t, temp in zip(temps, temperatures):
    print(t, temp)
```
